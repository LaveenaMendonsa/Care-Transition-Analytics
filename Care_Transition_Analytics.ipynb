## Exploratory Data Analysis

```python
df.describe()
```

```python
df=df.sort_values('Date')
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.plot(df['Date'], df['Children in CBP custody'])
plt.title('Daily Children in CBP Custody')
plt.xlabel('Date')
plt.ylabel('Number of Children')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Total Entries'] = df['Children apprehended and placed in CBP custody*']
```

```python
df['Total Exits'] = (
    df['Children transferred out of CBP custody']
    + df['Children discharged from HHS Care']
)
```

```python
df['Entry Exit Difference'] = (
    df['Total Entries'] - df['Total Exits']
)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(df['Date'], df['Total Entries'], label='Total Entries')
plt.plot(df['Date'], df['Total Exits'], label='Total Exits')
plt.title('Daily Entries vs Exits')
plt.xlabel('Date')
plt.ylabel('Number of Children / Stage Movements')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df[['Total Entries', 'Total Exits', 'Entry Exit Difference']].describe()
```

```python
df['Day'] = df['Date'].dt.day_name()
```

```python
df['Day Type'] = df['Date'].dt.dayofweek.apply(
    lambda x: 'Weekend' if x >= 5 else 'Weekday'
)
```

```python
df.groupby('Day Type')[
    ['Children transferred out of CBP custody',
     'Children discharged from HHS Care']
].mean()
```

```python
df.groupby('Day Type')[
    ['Children transferred out of CBP custody',
     'Children discharged from HHS Care']
].mean().plot(kind='bar', figsize=(9, 5))

plt.title('Average Transfers and Discharges: Weekday vs Weekend')
plt.xlabel('Day Type')
plt.ylabel('Average Number of Children')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df['Month'] = df['Date'].dt.to_period('M')
```

```python
monthly_trends = df.groupby('Month')[
    ['Children transferred out of CBP custody',
     'Children discharged from HHS Care']
].sum()

monthly_trends
```

```python
monthly_trends.plot(kind='line', marker='o', figsize=(12, 5))
plt.title('Monthly Transfer and Discharge Trends')
plt.xlabel('Month')
plt.ylabel('Total Number of Children')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Transfer Efficiency (%)'] = (
    df['Children transferred out of CBP custody']
    / df['Children in CBP custody']
) * 100
```

```python
df[['Date', 'Children in CBP custody',
    'Children transferred out of CBP custody',
    'Transfer Efficiency (%)']].head()
```

```python
df['Discharge Effectiveness (%)'] = (
    df['Children discharged from HHS Care']
    / df['Children in HHS Care']
) * 100
```

```python
df[['Date', 'Children in HHS Care',
    'Children discharged from HHS Care',
    'Discharge Effectiveness (%)']].head()
```

```python
total_entries = df['Total Entries'].sum()
total_exits = df['Total Exits'].sum()

throughput_rate = (total_exits / total_entries) * 100

print("Pipeline Throughput Rate:", throughput_rate)
```

```python
df['CBP Backlog Change'] = df['Children in CBP custody'].diff()
```

```python
df['HHS Backlog Change'] = df['Children in HHS Care'].diff()
```

```python
df[['Date', 'CBP Backlog Change', 'HHS Backlog Change']].head(10)
```

```python
discharge_mean = df['Children discharged from HHS Care'].mean()
discharge_std = df['Children discharged from HHS Care'].std()

cv = discharge_std / discharge_mean
stability_score = 100 / (1 + cv)

print("Discharge Mean:", discharge_mean)
print("Discharge Standard Deviation:", discharge_std)
print("Outcome Stability Score:", stability_score)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(df['Date'], df['CBP Backlog Change'])
plt.axhline(y=0, linestyle='--')
plt.title('Daily Change in CBP Custody')
plt.xlabel('Date')
plt.ylabel('Change in Children in Custody')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.plot(df['Date'], df['HHS Backlog Change'])
plt.axhline(y=0, linestyle='--')
plt.title('Daily Change in HHS Care')
plt.xlabel('Date')
plt.ylabel('Change in Children in HHS Care')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd

st.set_page_config(
    page_title="Care Transition Analytics",
    layout="wide"
)

st.title("Care Transition Efficiency & Placement Outcome Analytics")
st.write("Analyze CBP transfers, HHS care, discharges, and potential bottlenecks.")
```

```python
%%writefile -a app.py

import os

file_path = "HHS_Unaccompanied_Alien_Children_Program.csv"

if os.path.exists(file_path):
    df = pd.read_csv(file_path)

    df = df.dropna(how="all")
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

    df["Children in HHS Care"] = pd.to_numeric(
        df["Children in HHS Care"].astype(str).str.replace(",", "", regex=False),
        errors="coerce"
    )

    df = df.dropna(subset=["Date"])
    df = df.sort_values("Date")

    st.success("Dataset loaded successfully!")
    st.write("Dataset shape:", df.shape)
    st.dataframe(df.head())
else:
    st.error("CSV file not found. Place it in the same folder as app.py.")
```
